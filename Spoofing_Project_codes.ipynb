```python
#First_3 Model with advanced techinqes(The first Orginal code is the DNN)
# GNSS Spoofing Detection - Improved DNN (Keras) with advanced callbacks & safeguards
import os
import random
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime

import tensorflow as tf
from tensorflow.keras import regularizers
from tensorflow.keras.callbacks import (
    EarlyStopping, ReduceLROnPlateau, ModelCheckpoint,
    TensorBoard, CSVLogger
)

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_auc_score,
    roc_curve, auc
)
from sklearn.utils.class_weight import compute_class_weight


# Reproducibility seeds

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# -------------------------------
# 0) Config
# -------------------------------
DATA_FILE = "GNSS_raw_train.csv"
SCALER_PATH = "scaler_dnn.pkl"
BEST_MODEL_PATH = "gnss_deep_best.h5"
FINAL_MODEL_PATH = "gnss_deep_model.h5"
LOG_DIR = os.path.join("logs", "fit", datetime.now().strftime("%Y%m%d-%H%M%S"))

EPOCHS = 100
BATCH_SIZE = 32
LR = 1e-3
L2_WEIGHT = 1e-4  # L2 regularization coefficient
DROPOUT_RATES = (0.3, 0.2, 0.1)

# -------------------------------
# 1) Load & prepare data
# -------------------------------
if not os.path.exists(DATA_FILE):
    raise SystemExit(f"❌ File {DATA_FILE} not found. Place the CSV in the working directory.")

df = pd.read_csv(DATA_FILE)
df = df.dropna(subset=["Label"])
df["Label"] = df["Label"].astype(int)

# drop non-feature columns if present
X = df.drop(columns=["Label", "GPS_Time", "Latitude", "Longitude", "Height", "Satelite_Code"], errors="ignore")
y = df["Label"].values

print(f"Samples: {X.shape[0]} | Features: {X.shape[1]}")
print("Class distribution:\n", pd.Series(y).value_counts())

# Make sure labels are binary (0/1). If not, map them appropriately.
unique_labels = np.unique(y)
if set(unique_labels) - {0, 1}:
    print("Warning: labels are not {0,1}. Attempting to map them to 0/1 based on sorted unique values.")
    # simple mapping: smallest -> 0, others -> 1 (adjust if needed)
    label_map = {unique_labels[0]: 0}
    for i, lbl in enumerate(unique_labels[1:]):
        label_map[lbl] = 1
    y = np.vectorize(label_map.get)(y)
    print("Label mapping applied:", label_map)

# train / test split (stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=SEED, stratify=y
)

# scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

joblib.dump(scaler, SCALER_PATH)
print(f" Scaler saved as {SCALER_PATH}")

# compute class weights (helps with imbalance)
classes = np.unique(y_train)
cw = compute_class_weight(class_weight="balanced", classes=classes, y=y_train)
class_weight = {int(cls): float(w) for cls, w in zip(classes, cw)}
print("Class weights:", class_weight)

# -------------------------------
# 2) Build model
# -------------------------------
def build_model(input_dim):
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(input_dim,)),
        tf.keras.layers.Dense(256, activation="relu",
                              kernel_initializer="he_normal",
                              kernel_regularizer=regularizers.l2(L2_WEIGHT)),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Dropout(DROPOUT_RATES[0]),

        tf.keras.layers.Dense(128, activation="relu",
                              kernel_initializer="he_normal",
                              kernel_regularizer=regularizers.l2(L2_WEIGHT)),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Dropout(DROPOUT_RATES[1]),

        tf.keras.layers.Dense(64, activation="relu",
                              kernel_initializer="he_normal",
                              kernel_regularizer=regularizers.l2(L2_WEIGHT)),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Dropout(DROPOUT_RATES[2]),

        tf.keras.layers.Dense(32, activation="relu",
                              kernel_regularizer=regularizers.l2(L2_WEIGHT)),
        tf.keras.layers.BatchNormalization(),

        tf.keras.layers.Dense(1, activation="sigmoid")
    ])
    return model

input_dim = X_train_scaled.shape[1]
model = build_model(input_dim)

optimizer = tf.keras.optimizers.Adam(learning_rate=LR)
model.compile(
    optimizer=optimizer,
    loss="binary_crossentropy",
    metrics=["accuracy",
             tf.keras.metrics.AUC(name="auc"),
             tf.keras.metrics.Precision(name="precision"),
             tf.keras.metrics.Recall(name="recall")]
)

model.summary()

# -------------------------------
# 3) Callbacks
# -------------------------------
callbacks = []

# Early stopping
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True,
    verbose=1
)
callbacks.append(early_stop)

# Reduce LR on plateau
reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=5,
    min_lr=1e-6,
    verbose=1
)
callbacks.append(reduce_lr)

# Model checkpoint (save best by val_auc if available otherwise val_loss)
# ensure metric exists in history -> val_auc exists because we included AUC metric
checkpoint = ModelCheckpoint(
    BEST_MODEL_PATH,
    monitor="val_auc",
    mode="max",
    save_best_only=True,
    save_weights_only=False,
    verbose=1
)
callbacks.append(checkpoint)

# TensorBoard
tensorboard_cb = TensorBoard(log_dir=LOG_DIR, histogram_freq=1)
callbacks.append(tensorboard_cb)

# CSV logger
csv_logger = CSVLogger("training_log.csv", append=False)
callbacks.append(csv_logger)

# -------------------------------
# 4) Training
# -------------------------------
history = model.fit(
    X_train_scaled,
    y_train,
    validation_data=(X_test_scaled, y_test),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    class_weight=class_weight,
    callbacks=callbacks,
    verbose=1
)

# save final model (the best model is already saved by ModelCheckpoint)
model.save(FINAL_MODEL_PATH)
print(f" Final model saved as {FINAL_MODEL_PATH}")
print(f" Best model (by val_auc) saved as {BEST_MODEL_PATH}")

# -------------------------------
# 5) Plot training history
# -------------------------------
sns.set(style="whitegrid")
plt.figure(figsize=(14, 5))

# Accuracy
plt.subplot(1, 3, 1)
plt.plot(history.history.get("accuracy", []), label="train_acc")
plt.plot(history.history.get("val_accuracy", []), label="val_acc")
plt.title("Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

# Loss
plt.subplot(1, 3, 2)
plt.plot(history.history.get("loss", []), label="train_loss")
plt.plot(history.history.get("val_loss", []), label="val_loss")
plt.title("Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

# AUC
plt.subplot(1, 3, 3)
plt.plot(history.history.get("auc", []), label="train_auc")
plt.plot(history.history.get("val_auc", []), label="val_auc")
plt.title("AUC")
plt.xlabel("Epoch")
plt.ylabel("AUC")
plt.legend()

plt.tight_layout()
plt.show()

# -------------------------------
# 6) Evaluation on test set
# -------------------------------
y_pred_prob = model.predict(X_test_scaled).ravel()
y_pred = (y_pred_prob >= 0.5).astype(int)

print("\n📊 Classification Report:\n")
print(classification_report(y_test, y_pred, digits=4))

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Normal", "Spoofing"],
            yticklabels=["Normal", "Spoofing"])
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix")
plt.show()

# ROC and AUC
try:
    roc_auc = roc_auc_score(y_test, y_pred_prob)
    fpr, tpr, _ = roc_curve(y_test, y_pred_prob)
    roc_auc_val = auc(fpr, tpr)
    print(f"ROC AUC: {roc_auc:.4f}")

    plt.figure(figsize=(6,5))
    plt.plot(fpr, tpr, label=f"AUC = {roc_auc_val:.4f}")
    plt.plot([0,1], [0,1], linestyle="--", color="gray")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("ROC Curve")
    plt.legend()
    plt.grid(True)
    plt.show()
except Exception as e:
    print("Could not compute ROC AUC:", e)

# -------------------------------
# 7) Save scaler & (optionally) training history
# -------------------------------
joblib.dump(scaler, SCALER_PATH)
print(f" Scaler (again) saved as {SCALER_PATH}")

# Save history to CSV for later analysis
hist_df = pd.DataFrame(history.history)
hist_df.to_csv("training_history.csv", index=False)
print(" Training history saved to training_history.csv")

# End
print("Training and evaluation complete.")
```

```python
#Second Model
# GNSS Spoofing Detection - Random Forest Classifier

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.ensemble import RandomForestClassifier
import joblib


# 1) Load & prepare data

DATA_FILE = "GNSS_raw_train.csv"
if not os.path.exists(DATA_FILE):
    raise SystemExit(f" الملف {DATA_FILE} غير موجود.")

df = pd.read_csv(DATA_FILE)
df = df.dropna(subset=["Label"])
df["Label"] = df["Label"].astype(int)

X = df.drop(columns=["Label", "GPS_Time", "Latitude", "Longitude", "Height", "Satelite_Code"], errors="ignore")
y = df["Label"].values

print("Samples:", X.shape[0], "| Features:", X.shape[1])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

joblib.dump(scaler, "scaler.pkl")
print(" Scaler saved as scaler.pkl")


# 2) Train Random Forest

model = RandomForestClassifier(
    n_estimators=200,        # number of trees
    max_depth=12,            # tree depth
    random_state=42,
    n_jobs=-1,               # use all CPU cores
    class_weight="balanced"  # handle imbalance
)

print(" Training Random Forest...")
model.fit(X_train_scaled, y_train)

# -------------------------------
# 3) Evaluation
# -------------------------------
y_pred = model.predict(X_test_scaled)

print("\n Classification Report:\n")
print(classification_report(y_test, y_pred))

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=["Normal","Spoofing"],
            yticklabels=["Normal","Spoofing"])
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix")
plt.show()

# 4) Feature Importance
# -------------------------------
importances = model.feature_importances_
indices = np.argsort(importances)[::-1]

plt.figure(figsize=(10,6))
plt.bar(range(len(importances)), importances[indices], align="center")
plt.xticks(range(len(importances)), [X.columns[i] for i in indices], rotation=45)
plt.title("Feature Importance (Random Forest)")
plt.tight_layout()
plt.show()

# -------------------------------
# 5) Save model
# -------------------------------
joblib.dump(model, "gnss_rf_model.pkl")
print("✅ Model saved as gnss_rf_model.pkl")
```

```python
# third Model
# GNSS Spoofing Detection - XGBoost (Robust & Advanced Version)
# Includes: CV -> Final Training -> Keras-like Logs -> Plots -> Evaluation -> Save

import os
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import xgboost as xgb

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    classification_report, confusion_matrix,
    roc_auc_score, roc_curve, precision_recall_curve, average_precision_score
)


# Settings

DATA_FILE = "GNSS_raw_train.csv"
RANDOM_STATE = 42
N_FOLDS = 5
NUM_BOOST_ROUND = 1000
EARLY_STOP = 30
TEST_SIZE = 0.20

# 1) Load and Prepare Data

if not os.path.exists(DATA_FILE):
    raise SystemExit(f" File not found: {DATA_FILE}")

df = pd.read_csv(DATA_FILE)
df = df.dropna(subset=["Label"])
df["Label"] = df["Label"].astype(int)

X = df.drop(columns=["Label", "GPS_Time", "Latitude", "Longitude", "Height", "Satelite_Code"], errors="ignore")
y = df["Label"].values

print(f"Samples: {X.shape[0]} | Features: {X.shape[1]}")

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=y
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)
joblib.dump(scaler, "scaler_xgb.pkl")
print(" Scaler saved as scaler_xgb.pkl")


# 2) Cross Validation (xgb.cv)

dtrain_full = xgb.DMatrix(X_train_scaled, label=y_train)

params = {
    "objective": "binary:logistic",
    "eta": 0.05,
    "max_depth": 6,
    "subsample": 0.8,
    "colsample_bytree": 0.8,
    "seed": RANDOM_STATE,
    "verbosity": 1,
    "eval_metric": ["logloss", "error"]
}

print("\n Running xgb.cv to find optimal boosting rounds...")
cv_results = xgb.cv(
    params=params,
    dtrain=dtrain_full,
    num_boost_round=NUM_BOOST_ROUND,
    nfold=N_FOLDS,
    metrics=("logloss", "error"),
    early_stopping_rounds=EARLY_STOP,
    seed=RANDOM_STATE,
    verbose_eval=50,
    as_pandas=True
)

best_rounds = len(cv_results)
print(f" CV finished. Best rounds: {best_rounds}")
cv_results.to_csv("xgb_cv_results.csv", index=False)
print("Saved CV results to xgb_cv_results.csv")

# 3) Final Training

dtrain = xgb.DMatrix(X_train_scaled, label=y_train)
dtest  = xgb.DMatrix(X_test_scaled,  label=y_test)

evals_result = {}
print("\n Training final booster...")
booster = xgb.train(
    params=params,
    dtrain=dtrain,
    num_boost_round=best_rounds,
    evals=[(dtrain, "train"), (dtest, "eval")],
    evals_result=evals_result,
    verbose_eval=50
)

booster.save_model("gnss_xgb_booster.json")
print(" Booster saved as gnss_xgb_booster.json")


# 4) Keras-like Training Logs

train_logloss = evals_result.get("train", {}).get("logloss", [])
eval_logloss  = evals_result.get("eval",  {}).get("logloss", [])
train_error   = evals_result.get("train", {}).get("error", [])
eval_error    = evals_result.get("eval",  {}).get("error", [])

train_acc = [1.0 - e for e in train_error] if train_error else []
eval_acc  = [1.0 - e for e in eval_error]  if eval_error else []

rounds_ran = max(len(train_logloss), len(eval_logloss), 1)
print(f"\nRounds recorded: {rounds_ran}")

if train_acc and eval_acc:
    print("\n--- Training summary (Keras-like logs) ---")
    for i in range(rounds_ran):
        ta = train_acc[i] if i < len(train_acc) else np.nan
        tl = train_logloss[i] if i < len(train_logloss) else np.nan
        va = eval_acc[i]  if i < len(eval_acc)  else np.nan
        vl = eval_logloss[i]  if i < len(eval_logloss) else np.nan
        print(f"Epoch {i+1}/{rounds_ran} - accuracy: {ta:.4f} - loss: {tl:.4f} - val_accuracy: {va:.4f} - val_loss: {vl:.4f}")


# 5) Plot Training Curves

plt.figure(figsize=(14,5))

if train_acc and eval_acc:
    plt.subplot(1,2,1)
    plt.plot(range(1, len(train_acc)+1), train_acc, label="Train Accuracy")
    plt.plot(range(1, len(eval_acc)+1),  eval_acc,  label="Val Accuracy")
    plt.title("Model Accuracy")
    plt.xlabel("Round")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.grid(True)

plt.subplot(1,2,2)
if train_logloss and eval_logloss:
    plt.plot(range(1, len(train_logloss)+1), train_logloss, label="Train LogLoss")
    plt.plot(range(1, len(eval_logloss)+1),  eval_logloss,  label="Val LogLoss")
    plt.title("Model Loss (LogLoss)")
    plt.xlabel("Round")
    plt.ylabel("LogLoss")
    plt.legend()
    plt.grid(True)

plt.tight_layout()
plt.show()


# 6) Robust Prediction Function

def booster_predict_robust(booster_obj, dmatrix_obj):
    best_it = getattr(booster_obj, "best_iteration", None)
    best_ntree = getattr(booster_obj, "best_ntree_limit", None)

    attempts = []
    if best_ntree is not None:
        attempts.append(lambda: booster_obj.predict(dmatrix_obj, ntree_limit=best_ntree))
    if best_it is not None:
        attempts.append(lambda: booster_obj.predict(dmatrix_obj, ntree_limit=best_it+1))
        attempts.append(lambda: booster_obj.predict(dmatrix_obj, iteration_range=(0, best_it+1)))
        attempts.append(lambda: booster_obj.predict(dmatrix_obj, num_iteration=best_it))
    attempts.append(lambda: booster_obj.predict(dmatrix_obj))

    for fn in attempts:
        try:
            preds = fn()
            return np.array(preds)
        except Exception:
            continue
    raise RuntimeError("Unable to call booster.predict() with any known signature.")

# Predict
y_prob = booster_predict_robust(booster, dtest)
y_pred = (y_prob > 0.5).astype(int)

# 7) Evaluation

print("\n📊 Classification Report:\n")
print(classification_report(y_test, y_pred, digits=4))

cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=["Normal","Spoofing"], yticklabels=["Normal","Spoofing"])
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix (XGBoost)")
plt.tight_layout()
plt.show()

# ROC & PR
auc = roc_auc_score(y_test, y_prob)
precision, recall, _ = precision_recall_curve(y_test, y_prob)
ap = average_precision_score(y_test, y_prob)
fpr, tpr, _ = roc_curve(y_test, y_prob)

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(fpr, tpr, label=f"AUC = {auc:.4f}")
plt.plot([0,1],[0,1],"--", color="gray")
plt.title("ROC Curve")
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.legend()
plt.grid(True)

plt.subplot(1,2,2)
plt.plot(recall, precision, label=f"AP = {ap:.4f}")
plt.title("Precision-Recall Curve")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# 8) Feature Importance

try:
    ax = xgb.plot_importance(booster, max_num_features=15, height=0.6)
    ax.figure.set_size_inches(10,6)
    plt.title("Feature importance (XGBoost)")
    plt.show()
except Exception as e:
    print(" Couldn't plot feature importance:", e)


# 9) Save Predictions

out_df = pd.DataFrame({"y_true": y_test, "y_prob": y_prob, "y_pred": y_pred})
out_df.to_csv("xgb_test_predictions.csv", index=False)
booster.save_model("gnss_xgb_booster.json")   # ensure final save
print(" Predictions saved to xgb_test_predictions.csv")
print(" Booster saved as gnss_xgb_booster.json")
print("\n XGBoost run complete (robust & advanced).")
```
